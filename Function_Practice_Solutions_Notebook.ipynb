```
def lesser_of_two_evens(a,b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a,b)
    return max(a,b)

lesser_of_two_evens(2,4)
```

```
def animal_crackers(text):
    w1, w2 = text.lower().split()
    return w1[0] == w2[0]

animal_crackers('Levelheaded Llama')
```

```
def makes_twenty(n1,n2):
    return n1 == 20 or n2 == 20 or n1+n2 == 20

makes_twenty(20,10)
```

```
def old_macdonald(name):
    return name[:3].capitalize() + name[3:].capitalize()

old_macdonald('macdonald')
```

```
def master_yoda(text):
    return " ".join(text.split()[::-1])

master_yoda('I am home')
```

```
def almost_there(n):
    return abs(100-n) <= 10 or abs(200-n) <= 10

almost_there(104)
```

```
def has_33(nums):
    for i in range(len(nums)-1):
        if nums[i] == nums[i+1] == 3:
            return True
    return False

has_33([1,3,3])
```

```
def paper_doll(text):
    return ''.join(ch*3 for ch in text)

paper_doll('Hello')
```

```
def blackjack(a,b,c):
    total = a+b+c
    if total <= 21:
        return total
    if 11 in [a,b,c]:
        total -= 10
        if total <= 21:
            return total
    return 'BUST'

blackjack(9,9,11)
```

```
def summer_69(arr):
    total = 0
    ignore = False
    for num in arr:
        if num == 6:
            ignore = True
        elif num == 9 and ignore:
            ignore = False
        elif not ignore:
            total += num
    return total

summer_69([2,1,6,9,11])
```

```
def spy_game(nums):
    code = [0,0,7]
    for num in nums:
        if code and num == code[0]:
            code.pop(0)
    return len(code) == 0

spy_game([1,2,4,0,0,7,5])
```

```
def count_primes(num):
    if num < 2:
        return 0
    count = 0
    for n in range(2, num+1):
        prime = True
        for i in range(2, int(n**0.5)+1):
            if n % i == 0:
                prime = False
                break
        if prime:
            count += 1
    return count

count_primes(100)
```

```
def print_big(letter):
    patterns = {
        1:'  *  ',
        2:' * * ',
        3:'*****',
        4:'*   *',
        5:'*    ',
        6:'**** '
    }

    alphabet = {
        'A':[1,2,3,4,4],
        'B':[6,4,6,4,6],
        'C':[3,5,5,5,3],
        'D':[6,4,4,4,6],
        'E':[3,5,3,5,3]
    }

    for p in alphabet[letter.upper()]:
        print(patterns[p])

print_big('A')
```
